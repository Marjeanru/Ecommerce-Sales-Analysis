```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns 
```

```python
sns.set_theme(style='whitegrid')
```

```python
CA2=pd.read_csv('../data/CA2.csv')
```

```python
CA2
```

```python
CA2['total_spend']=CA2['total_spend'].str.replace('€','').str.replace(',','').astype(float)
```

```python
# I NOTICED THE DATA HAS SOME ERRORS SO I CORRECTED THEM. DUPLICATED COUNTRIES NAMES. 

CA2['country']= CA2['country'].replace('España','Spain').replace('Deutschland','Germany')
```

```python
CA2=CA2.groupby('country',as_index=False)['total_spend'].sum()
```

```python
CA2=CA2.sort_values(by='total_spend')
```

```python
CA2
```

```python
import matplotlib.ticker as ticker 

plt.figure(figsize=(10,6))
sns.barplot(x='country',y='total_spend',data=CA2)
plt.xticks(rotation=45)
plt.gca().yaxis.set_major_formatter(ticker.StrMethodFormatter('{x:,.0f}'))
plt.gca().yaxis.set_major_locator(ticker.MaxNLocator(nbins=10))
plt.xlabel('COUNTRY',labelpad=20,fontsize= 15)
plt.ylabel('TOTAL SPEND',labelpad=20,fontsize= 15)
```
