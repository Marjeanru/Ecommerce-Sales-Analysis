```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns 
```

```python
sns.set_theme(style='whitegrid')
```

```python
UJ1=pd.read_csv('../data/UJ1.csv')
```

```python
UJ1
```

```python
import matplotlib.ticker as ticker 

plt.figure(figsize=(10,6))
Source=sns.barplot(y='traffic_source',x='times_source_used', data=UJ1,color='purple')
plt.xlabel('Times Used', fontsize=15,fontweight='bold',labelpad= 16)
plt.ylabel('Traffic Source', fontsize=15,fontweight='bold',labelpad= 16)
Source.bar_label(Source.containers[0], fmt='{:,.0f}',padding=5, fontweight='bold',rotation=270)
plt.gca().xaxis.set_major_locator(ticker.MaxNLocator(nbins=10))
```

```python
UJ2=pd.read_csv('../data/UJ2.csv')
```

```python
UJ2
```

```python
UJ2['event_type']=UJ2['event_type'].str.capitalize()
```

```python
#Next graphic shows where the search end up

plt.figure(figsize=(11,6))
Session=sns.barplot(y='event_type',x='total_sessions',data=UJ2,color='darkblue')
plt.xlabel('Total Sessions', fontsize=15,fontweight='bold',labelpad= 16)
plt.ylabel('Event Type', fontsize=15,fontweight='bold',labelpad= 16)
Session.bar_label(Session.containers[0], fmt='{:,.0f}', padding=5, fontweight='bold',rotation=270)
plt.show()
```
