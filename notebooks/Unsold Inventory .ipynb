```python
hmap_UI3=UI3.set_index('product_category').sort_values(by='unsold_items',ascending=False)
```

```python
plt.figure(figsize=(10,7))
sns.heatmap(data=hmap_UI3,cmap='viridis',annot=True,fmt=',d')
plt.title('UNSOLD ITEMS BY PRODUCT_CATEGORY',fontsize=13,fontweight='bold',pad=20)
plt.ylabel('')
```
