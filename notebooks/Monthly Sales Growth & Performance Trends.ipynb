```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.dates as mdate
import seaborn as sns 
    
```

```python
sns.set_theme(style='whitegrid')
```

```python
MSG1=pd.read_csv('../data/MSG1.csv')
```

```python
MSG1
```

```python
MSG1['date']=pd.to_datetime(MSG1['month']+' '+MSG1['year'].astype(str),format='%B %Y')
```

```python
MSG1=MSG1.drop(columns=['month','year'])
MSG1['total_sales_month'] = (MSG1['total_sales_month'].astype(str).str.replace('€', '').str.replace(' ', '').astype(float))
```

```python
plt.figure(figsize=(12,8))
sns.lineplot(x='date',y='total_sales_month',data=MSG1,color='r')
plt.yticks(np.linspace(MSG1['total_sales_month'].min(),MSG1['total_sales_month'].max(),15))
plt.gca().xaxis.set_major_formatter(mdate.DateFormatter('%B %Y'))
plt.gca().xaxis.set_major_locator(mdate.MonthLocator(interval=3))
plt.xticks(rotation=45)
plt.ylabel('Total Sales €',labelpad=25,fontsize=13)
plt.xlabel('Date',labelpad=25,fontsize=13)
plt.title('Monthly Performance',pad=30,fontsize=15)
```
