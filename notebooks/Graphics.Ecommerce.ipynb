```python
EFS3=pd.read_csv('../data/EFS3.csv')
```

```python
EFS3['revenue_lost']=(EFS3['revenue_lost'].str.replace(',', '').str.replace('€','').astype(float))
```

```python
EFS3=EFS3.sort_values(by='revenue_lost')
```

```python
plt.figure(figsize=(11,8))
sns.barplot(y='revenue_lost',hue='category', data=EFS3, palette='rainbow')
plt.legend(bbox_to_anchor= (1.4,1.01),loc='best', fontsize='medium')
plt.yticks(fontsize=15)
plt.ylabel('revenue_lost',fontsize=20,labelpad=20)

```
